### Practical 1: Count Lines, Words and Characters

**Problem Definition:**  
Write a Python program to read text from a file and count the total number of lines, words, and characters. The result should be written to an output file.


```python
with open("file1.txt","r") as fp:
    txt = fp.read()

lines = txt.split("\n")
lines_count = len(lines)

words = txt.split()
words_count = len(words)

char = len(txt)

with open("output.txt","w") as fp1:
    fp1.write(f"No of Lines:{lines_count}\n")
    fp1.write(f"No of Words:{words_count}\n")
    fp1.write(f"No of Char:{char}\n")
```


### Practical 2: Search a Word in a File

**Problem Definition:**  
Write a Python program to search a given word in a text file and count its occurrences. Store the result in an output file.


```python
search_word = input("Enter Word to Search:")

with open('file1.txt','r') as fp:
    txt = fp.read()

txt = txt.lower()
search_word = search_word.lower()  

word_count = txt.split().count(search_word)

with open('output.txt','w') as fp1:

    fp1.write(f"Words Searched:{search_word}\n")
    fp1.write(f"Occurence:{word_count}")
```


### Practical 3: Student Marks and Average

**Problem Definition:**  
Write a Python program to read roll number, name, and marks of students from a file. Store marks in a list, calculate the average using a user-defined function, and write student details along with average marks to an output file.


```python
def calculate_average(marks):
    return sum(marks) / len(marks)

with open("new.txt", "r") as infile:
    lines = infile.readlines()

with open("output.txt", "w") as outfile:
    
    for line in lines:

        data = line.strip().split(",")
        print(data)
        
        roll = data[0]
        name = data[1]
        
        marks = list(map(int, data[2:]))
        
        avg = calculate_average(marks)
        
        outfile.write(f"Roll No: {roll}, Name: {name}, Marks: {marks}, Average: {avg:.2f}\n")

print("Data processed successfully. Check output.txt")
```


### Practical 4: Maximum and Minimum using Tuple

**Problem Definition:**  
Write a Python program to read numbers from a file, store them in a tuple, and find the maximum and minimum values. Write the result to an output file.


```python
with open("new.txt", "r") as file:
    content = file.read()

numbers = tuple(map(int, content.split(",")))

maximum = max(numbers)
minimum = min(numbers)

with open("output1.txt", "w") as file:
    file.write("Numbers: " + str(numbers) + "\n")
    file.write("Maximum Value: " + str(maximum) + "\n")
    file.write("Minimum Value: " + str(minimum) + "\n")

print("Result written to output1.txt successfully!")
```


### Practical 5: Remove Duplicate Words using Set

**Problem Definition:**  
Write a Python program to read words from a file and remove duplicate words using a set. Store the unique words in an output file.


```python
with open("new.txt", "r") as file:
    content = file.read()


words = content.split(",")

unique_words = set(words)

with open("output5.txt", "w") as file:
    file.write("Unique Words:\n")
    for word in unique_words:
        file.write(word + "\n")

print("Duplicate words removed and stored in output5.txt successfully!")
```


### Practical 6: Word Frequency using Dictionary

**Problem Definition:**  
Write a Python program to count the frequency of each word in a text file using a dictionary and write the result to an output file.


```python
with open("input6.txt", "r") as file:
    content = file.read()

content = content.lower()

words = content.split()

word_freq = {}

for word in words:
    if word in word_freq:
        word_freq[word] += 1
    else:
        word_freq[word] = 1

with open("output6.txt", "w") as file:
    file.write("Word Frequencies:\n")
    for word, count in word_freq.items():
        file.write(f"{word} : {count}\n")

print("Word frequency written to output6.txt successfully!")
```


### Practical 7: Student Result and Grade Calculation

**Problem Definition:**  
Write a Python program to read roll number, name, and marks of students from a file. Use a user-defined function to determine grade (A/B/C/Fail) and write the student result to an output file.

**Grade Criteria:**  
- Marks ≥ 75 : Grade A  
- Marks ≥ 60 : Grade B  
- Marks ≥ 40 : Grade C  
- Marks < 40 : Fail


```python
def calculate_grade(marks):
    if marks >= 75:
        return "A"
    elif marks >= 60:
        return "B"
    elif marks >= 40:
        return "C"
    else:
        return "Fail"

with open("input7.txt", "r") as file:
    lines = file.readlines()

with open("output7.txt", "w") as output:
    output.write("Student Results:\n")
    output.write("----------------------------\n")
    
    for line in lines:
        roll, name, marks = line.strip().split(",")
        marks = float(marks)
        
        grade = calculate_grade(marks)
        
        output.write(f"Roll No: {roll}, Name: {name}, Marks: {marks}, Grade: {grade}\n")

print("Student results written to output7.txt successfully!")
```
