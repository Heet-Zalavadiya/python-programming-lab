### 10) WAP to test if the given tuple is distinct or not.
**Sample Input:**
```
tuple1 = (1, 2, 3, 4)
tuple2 = (1, 2, 2, 3)
```
**Sample Output:**
```
(1, 2, 3, 4) is distinct? -> True
(1, 2, 2, 3) is distinct? -> False
```

````python
tuple1 = (1, 2, 3, 4)
tuple2 = (1, 2, 2, 3)
L1 = list(tuple1)
L2 = list(tuple2)
l1 = list(set(tuple1))
l2 = list(set(tuple2))

if l2 == (L2):
    print(f"{tuple2} is distinct-> True")
else:
   print(f"{tuple2} is distinct-> False")


if l1 == (L1):
    print(f"{tuple1} is distinct-> True")
else:
   print(f"{tuple1} is distinct-> False")

````

### 11) WAP to rotate the elements of a tuple to the right by `k` positions.

```

```

