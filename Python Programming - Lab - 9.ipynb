### 02) Write a function that add first n numbers.

```python
def add(n:int) -> int:
    if(n == 0):
        return 0
    return n + add(n-1)
print(add(10))
```

### 03) Write a function that returns 1 if the given number is Prime or 0 otherwise.

```python
def prime(n):
    for i in range(2,n-1) :
        if(n%i == 0):
            return 0
    return 1

print(prime(7))
    
```

### 04) Write a function that returns the list of Prime numbers between given two numbers.

```python
li = []
def prime(n):
    for i in range(2,n-1) :
        if(n%i == 0):
            return 0
    li.append(i)
    return 1
    
def print_prime(start,end):
    for i in range(start,end+1):
        prime(i)

print_prime(10,20)
print(li)
```

### 06) Write a function that returns the sum of all the elements of the list.

### 07) Write a function to calculate the sum of the first element of each tuples inside the list.

```python
def addfirst(li:list)->int:
    sum = 0
    for i in li:
        sum += i[0]
    return sum
print(addfirst([(10,20,30),(20,30,40),(30,40,50)]))
```

### 08) Write a recursive function to find nth term of Fibonacci Series.

```python
def fibbo(n:int)->int:
    if n <= 0:
        return -1
    if n == 1:
        return 0
    if n == 2:
        return 1
    return fibbo(n-1) + fibbo(n-2)
print(fibbo(10))
```

### 11) Write a function to invert a given Dictionary.
#### hint: keys to values & values to keys
#### Before : {'a': 10, 'b':20, 'c':30, 'd':40}
#### After : {10:'a', 20:'b', 30:'c', 40:'d'}

```python
def invertDict(d:dict)->dict:
    d1 = {}
    for key,val in d.items():
        d1[val] = key
    return d1
d = {'a': 10, 'b':20, 'c':30, 'd':40}
print(invertDict(d))
```

### 12) Write a function that returns the number of uppercase and lowercase letters in the given string. 
#### example : Input : s1 = AbcDEfgh ,Ouptput : no_upper = 3, no_lower = 5

```python
def uplp(s1:str)->int:
    low = "qwertyuiopasdfghjklzxcvbnm"
    up = "QWERTYUIOPASDFGHJKLZXCVBNM"
    no_low = 0
    no_up = 0
    for i in low:
        no_low += s1.count(i)
    for i in up:
        no_up += s1.count(i)
    print(f"no_upper = {no_up},no_lower = {no_low}")
s1 = "AbcDEfgh"
uplp(s1)
```
